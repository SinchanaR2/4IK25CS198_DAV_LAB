```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

load the dataset


```python

df= pd.read_csv("/content/student_data.csv")
print("Dataset shape:",df.shape)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

1.Histogram disribution of final grade


```python

plt.figure(figsize=(6, 4))
sns.histplot(df["G3"],bins=20,kde=True,color="steelblue")
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

2.Box plot - grade spread across study time levels


```python

plt.figure(figsize=(6, 4))
sns.boxplot(x= "studytime" ,y="G3" ,data=df ,palette="Set2")
plt.title("Final Grade across Study time Levels")
plt.xlabel("Study Time (1=low, 4=high)")
plt.ylabel("Final Grade")
plt.show()
```

3.Scatter plot- relationship between two grade periods


```python

plt.figure(figsize=(6, 4))
sns.scatterplot(x= "G1", y="G3", hue="sex", data=df, palette="coolwarm")
plt.title("First Period Grade is Final Grade")
plt.xlabel("G1 (Period 1 Grade)")
plt.ylabel("G3 (Final Grade)")
plt.show()
```

4.Pair plot - pairwise relationship across key numeric values

```python
sns.pairplot(df[["age" , "studytime" , "G1" , "G2" , "G3" , "sex"]], hue="sex",palette ="husl")
plt.suptitle("Pairwise relationship",y=1.82)
plt.show()
```

5. Correlation Headmap - strength of relationship between numerical features

```python
numeric_cols = ["age", "studytime", "failures", "absences", "G1", "G2", "G3"]
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot =True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
