```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

STEP 1: Load the dataset

```python
df=pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 rows of dataset:")
print(df.head())
```

```python
print("\nDataset Info:")
print(df.info())
```

STEP 2: Check the missing values

```python
print("\nMissing values before handling:")
print(df.isnull().sum())
```

```python
df.loc[0:5, 'Study_Hours_per_Week']=None
df.loc[10:15, 'Attendance_Rate']=None
df.loc[20:15, 'Past_Exam_Scores']= None
df.loc[30:35, 'Final_Exam_Score']= None

print("\nMissing values after inserting sample NaNs:")
print(df.isnull().sum()[df.isnull().sum()>0])
```

STEP 3: Handle missing data using Different imputation methods

Mean Imputation

```python
df['Study_Hours_per_Week']= df['Study_Hours_per_Week'].fillna(df['Study_Hours_per_Week'].mean())
```

Median Imputation

```python
df['Attendance_Rate'] = df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

Forward Fill

```python
df['Past_Exam_Scores']= df['Past_Exam_Scores'].ffill()
```

Missing values after imputation

```python
df['Final_Exam_Score']= df['Final_Exam_Score'].fillna(df['Final_Exam_Score'].mean())
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

STEP 4: Encoding categorical variables

One-Hot Encoding

```python
# The 'Parental_Education_Level' column has already been one-hot encoded.
# The original column is no longer present in the DataFrame, as it was replaced by new columns (e.g., 'Parental_Education_Level_PhD').
# Attempting to re-encode it would result in a KeyError.

print("\nOne-Hot Encoded columns for 'Parental_Education_Level':")
encoded_columns = [col for col in df.columns if col.startswith('Parental_Education_Level_')]
if encoded_columns:
    print(encoded_columns)
else:
    print("No columns starting with 'Parental_Education_Level_' were found. The encoding might not have been applied or completed.")

print("\nAll current columns in the DataFrame:")
print(df.columns.tolist())
```

```python
le = LabelEncoder()
df['Pass_Fail_Encoded'] = le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Result:")
print(df[['Pass_Fail', 'Pass_Fail_Encoded']].head())
```

Label Encoding for other Binarycolumns too

```python
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Encoded'] = le.fit_transform(df['Extracurricular_Activities'])
```

STEP 5: Save the final cleaned dataset

```python
df.to_csv("student_performance_cleaned.csv", index = False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
